```python
import os
import glob
import json
import time

from dotenv import load_dotenv

import cv2
import numpy as np
    
from koger_detection.utils.video import extract_frames


load_dotenv()  # Loads .env file
```

```python
os.getenv("VIDEO_FOLDER")
```

```python
# Path to folder where videos are saved
video_folder_path = os.getenv("VIDEO_FOLDER")

video_files = glob.glob(os.path.join(video_folder_path, "to-annotate", "**", "*.mp4"), recursive=True)


out_folder = os.path.join(os.getenv("VIDEO_FOLDER"), "annotations", "extracted_frames")
os.makedirs(out_folder, exist_ok=True)
```

```python
for video_file in video_files[1:]:
    frame_out_folder = os.path.join(out_folder, video_file.split("/")[-2], os.path.splitext(video_file.split("/")[-1])[0])
    os.makedirs(frame_out_folder, exist_ok=True)
    extract_frames(video_file, frame_out_folder, skip=6)
```
